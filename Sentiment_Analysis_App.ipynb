```python
import gradio as gr
from textblob import TextBlob

def analyze_sentiment(text):
    if not text.strip():
        return "Please enter some text.", "Neutral", 0.0

    blob = TextBlob(text)
    polarity = blob.sentiment.polarity

    if polarity > 0.1:
        sentiment = "Positive 😊"
    elif polarity < -0.1:
        sentiment = "Negative 😠"
    else:
        sentiment = "Neutral 😐"

    return text, sentiment, round(polarity, 3)

# Gradio Interface
demo = gr.Interface(
    fn=analyze_sentiment,
    inputs=gr.Textbox(lines=4, placeholder="Enter your text here..."),
    outputs=[
        gr.Textbox(label="Your Text"),
        gr.Textbox(label="Sentiment"),
        gr.Number(label="Polarity Score (-1 to 1)")
    ],
    title="Sentiment Analysis App",
    description="Enter any text and find out if it's Positive, Negative, or Neutral.",
    theme="soft"
)

demo.launch(share=True)
```

```python
import gradio as gr
from textblob import TextBlob

def analyze_sentiment(text):
    if not text.strip():
        return "Please enter some text.", "Neutral 😐", 0.0, "Waiting for input..."

    blob = TextBlob(text)
    polarity = blob.sentiment.polarity

    if polarity > 0.2:
        sentiment = "Positive 😊"
        meaning = "The text expresses positive emotions or opinions."
    elif polarity < -0.2:
        sentiment = "Negative 😠"
        meaning = "The text expresses negative emotions or opinions."
    else:
        sentiment = "Neutral 😐"
        meaning = "The text is mostly neutral or mixed."

    return text, sentiment, round(polarity, 3), meaning

demo = gr.Interface(
    fn=analyze_sentiment,
    inputs=gr.Textbox(
        lines=5,
        placeholder="Type or paste your text here...\nExample: I really loved the movie!",
        label="Enter Text"
    ),
    outputs=[
        gr.Textbox(label="Your Text"),
        gr.Textbox(label="Sentiment"),
        gr.Number(label="Polarity Score (-1 to +1)"),
        gr.Textbox(label="Explanation")
    ],
    title="AI Sentiment Analysis App",
    description="This app uses Natural Language Processing (NLP) to detect whether the given text is Positive, Negative, or Neutral.",
    examples=[
        ["I absolutely love this product! It's amazing."],
        ["This is the worst service I have ever experienced."],
        ["The weather is okay today, nothing special."],
        ["I'm so happy and excited for the weekend!"]
    ]
)

demo.launch(share=True)
```
